```python
# Part 1: Native Python Data Structures & Comprehensions
# List Transformations

heights_in = [75, 82, 78, 69, 85, 71, 79]

heights_m = [round(height * 0.0254, 2) for height in heights_in]
heights_greater_76 = [height for height in heights_in if height > 76]

print("Heights in inches:", heights_in)
print("Heights in meters:", heights_m)
print("Heights greater than 76 inches:", heights_greater_76)

player_scores = {
 'LeBron James': 25.0,
 'Stephen Curry': 26.4,
 'Giannis Antetokounmpo': 28.1,
 'Jayson Tatum': 26.9,
 'Chris Paul': 13.9,
 'Marcus Smart': 11.5
}

player_scores_above_20 = {player: score for player, score in player_scores.items() if score > 20}
print("Player scores above 20:", player_scores_above_20)

average_ppg = sum(player_scores.values()) / len(player_scores)
print("Average points per game:", round(average_ppg, 2))

player_scores['Luka Dončić'] = 28.4

player_data = [
 ('Player A', 'PG', 22),
 ('Player B', 'SG', 18),
 ('Player C', 'PG', 29),
 ('Player D', 'SF', 15),
 ('Player E', 'SG', 24),
 ('Player F', 'SF', 12)
]

def total_points_per_position(data):
    position_points = {}
    for player, position, points in data:
        if position not in position_points:
            position_points[position] = 0
        position_points[position] += points
    return position_points

print("Total points per position:", total_points_per_position(player_data))

def average_points_per_position(data):
    position_points = {}
    position_counts = {}
    for player, position, points in data:
        if position not in position_points:
            position_points[position] = 0
            position_counts[position] = 0
        position_points[position] += points
        position_counts[position] += 1
    return {position: round(position_points[position] / position_counts[position], 2) for position in position_points}

print("Average points per position:", average_points_per_position(player_data))
```
