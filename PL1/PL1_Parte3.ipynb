```python
import pandas as pd

data = {
    'Name': ['A. Davis', 'J. Butler', 'K. Leonard', 'J. Embiid', 'D. Lillard'],
    'Team': ['LAL', 'MIA', 'LAC', 'PHI', 'MIL'],
    'Position': ['PF', 'SF', 'SF', 'C', 'PG'],
    'Age': [30, 34, 32, 29, 33],
    'Salary': [40600000, 45180000, 45640000, 46900000, 45640000]
}

df = pd.DataFrame(data)

df.info()
print()

name_salary = df[['Name', 'Salary']]
print("Name and Salary:\n", name_salary)
print()

rows_1_to_3 = df.iloc[1:4]
print("Rows 1 to 3:\n", rows_1_to_3)
print()

embiid_salary = df.loc[df['Name'] == 'J. Embiid', 'Salary'].values[0]
print("J. Embiid's Salary:", embiid_salary)

filtered_df = df[(df['Age'] > 30) & (df['Salary'] > 45000000)]
print("Filtered DataFrame:\n", filtered_df)
print()

df['Salary_Millions'] = df['Salary'] / 1_000_000
print("DataFrame with Salary in Millions:\n", df)
print()

df.to_csv('nba_players.csv', index=False)   

df_imported = pd.read_csv('nba_players.csv')
print("Imported DataFrame from CSV:\n", df_imported)
```
