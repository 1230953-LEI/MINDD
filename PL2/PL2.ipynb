```python
import numpy as np

def convert_to_cm(height):
    if pd.isna(height):
        return np.nan
    feet, inches = map(int, height.split('-'))
    return round((feet * 12 + inches) * 2.54, 2)

df['Height'] = df['Height'].apply(convert_to_cm)

print("Novo tipo da coluna 'Height':", df['Height'].dtype)
display(df[['Name', 'Height']].head())
```

```python
# Contar os valores nulos (NaN) em cada coluna
df.isna().sum()
```

```python
df = df.dropna(how='all')
print("Dimensões após remover linhas com todos os valores nulos:", df.shape)
```

```python
df['College'] = df['College'].fillna('No College')

print("Valores nulos em College após preenchimento:", df['College'].isna().sum())

df[df['College'] == 'No College'][['Name','Team','College']].head()
```

```python
salary_nan = df[df['Salary'].isna()]

salary_nan[['Name', 'Team', 'Salary']]
```

```python
# 1. Identificar e guardar os índices das linhas que têm salário nulo
indices_nan = df[df['Salary'].isna()].index

print("--- Antes da Imputação ---")
print("Total de NaNs na coluna Salary:", df['Salary'].isna().sum())

# 2. Imputação de nível 1: Média da respetiva Equipa e Posição
df['Salary'] = df['Salary'].fillna(df.groupby(['Team', 'Position'])['Salary'].transform('mean'))

print("\n--- Após 1º Nível (Média por Team e Position) ---")
print("Total de NaNs restantes na coluna Salary:", df['Salary'].isna().sum())

# 3. Imputação de nível 2: Média global da Posição (caso sobrasse algum nulo)
df['Salary'] = df['Salary'].fillna(df.groupby('Position')['Salary'].transform('mean'))

print("\n--- Após 2º Nível (Média Global por Position) ---")
print("Total de NaNs restantes na coluna Salary:", df['Salary'].isna().sum())

# 4. Apresentar os jogadores cujos valores foram imputados (formatado em dólares)
print("\n--- Jogadores e novos salários atribuídos ---")
pd.options.display.float_format = '${:,.2f}'.format
display(df.loc[indices_nan, ['Name', 'Team', 'Position', 'Salary']])
```

```python
# 1. Contagem detalhada de nulos por coluna
print("--- Nulos por Coluna ---")
print(df.isna().sum())

total_nans = df.isna().sum().sum()
print("\n--- Total Global de Nulos no DataFrame ---")
print(f"Total de NaNs restantes: {total_nans}")

# 3. Verificação lógica / asserção
if total_nans == 0:
    print("\n Sucesso: O DataFrame está completamente limpo e sem valores nulos!")
else:
    print(f"\n Atenção: Ainda existem {total_nans} valores nulos por tratar.")
```

```python
# Calcular a média das duas colunas
mean_stats = df[['Age', 'Salary']].mean()

print(mean_stats)
```

```python
# 1. Calcular as médias de referência
avg_age = df['Age'].mean()
avg_salary = df['Salary'].mean()

print(f"Critérios: Idade < {avg_age:.2f} anos e Salário > ${avg_salary:,.2f}\n")

# 2. Filtrar jogadores que cumprem ambas as condições
young_and_rich = df[(df['Age'] < avg_age) & (df['Salary'] > avg_salary)]

# 3. Ordenar por Age e Salary (por exemplo, Idade crescente e Salário decrescente)
sorted_young_rich = young_and_rich.sort_values(
    by=['Age', 'Salary'],
    ascending=[True, False]
)[['Name', 'Team', 'Position', 'Age', 'Salary']]

# 4. Apresentar o número total e os primeiros registos
print(f"Total de jogadores encontrados: {len(sorted_young_rich)}")
display(sorted_young_rich)
```

```python
# 1. Calcular a média salarial por equipa
team_avg_salary = df.groupby('Team')['Salary'].mean()

# 2. Equipa com o maior salário médio
highest_team = team_avg_salary.idxmax()
highest_val = team_avg_salary.max()

# 3. Equipa com o menor salário médio
lowest_team = team_avg_salary.idxmin()
lowest_val = team_avg_salary.min()

print("--- Equipa com Maior e Menor Salário Médio ---")
print(f"Maior média: {highest_team} com ${highest_val:,.2f}")
print(f"Menor média: {lowest_team} com ${lowest_val:,.2f}")

# 4. As três equipas com maior salário médio
top3_teams = team_avg_salary.nlargest(3)

print("\n--- Top 3 Equipas com Maior Média Salarial ---")
for rank, (team, salary) in enumerate(top3_teams.items(), 1):
    print(f"{rank}. {team}: ${salary:,.2f}")
```

```python
# Obter o índice da linha com maior salário em cada equipa
idx_max_salary = df.groupby('Team')['Salary'].idxmax()

# Selecionar essas linhas completas no DataFrame original
top_player_per_team = df.loc[idx_max_salary, ['Team', 'Name', 'Position', 'Salary']].sort_values(by='Team')

display(top_player_per_team)
```

```python
# 1. Resumo estatístico completo do salário por posição
salary_stats_by_pos = df.groupby('Position')['Salary'].describe()

# Formatar a apresentação dos números da tabela
pd.options.display.float_format = '${:,.2f}'.format
print("--- Estatísticas de Salário por Posição ---")
display(salary_stats_by_pos)

# 2. Identificar a posição com o maior salário médio
mean_salaries = df.groupby('Position')['Salary'].mean()
highest_pos = mean_salaries.idxmax()
highest_pos_val = mean_salaries.max()

print(f"\nPosição com maior salário médio: {highest_pos} (${highest_pos_val:,.2f})")
```
