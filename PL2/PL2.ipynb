```python
import numpy as np

def convert_to_cm(height):
    if pd.isna(height):
        return np.nan
    feet, inches = map(int, height.split('-'))
    return round((feet * 12 + inches) * 2.54, 2)

df['Height'] = df['Height'].apply(convert_to_cm)

print("Novo tipo da coluna 'Height':", df['Height'].dtype)
display(df[['Name', 'Height']].head())
```

```python
# 1. Identificar e guardar os índices das linhas que têm salário nulo
indices_nan = df[df['Salary'].isna()].index

print("--- Antes da Imputação ---")
print("Total de NaNs na coluna Salary:", df['Salary'].isna().sum())

# 2. Imputação de nível 1: Média da respetiva Equipa e Posição
df['Salary'] = df['Salary'].fillna(df.groupby(['Team', 'Position'])['Salary'].transform('mean'))

print("\n--- Após 1º Nível (Média por Team e Position) ---")
print("Total de NaNs restantes na coluna Salary:", df['Salary'].isna().sum())

# 3. Imputação de nível 2: Média global da Posição (caso sobrasse algum nulo)
df['Salary'] = df['Salary'].fillna(df.groupby('Position')['Salary'].transform('mean'))

print("\n--- Após 2º Nível (Média Global por Position) ---")
print("Total de NaNs restantes na coluna Salary:", df['Salary'].isna().sum())

# 4. Apresentar os jogadores cujos valores foram imputados (formatado em dólares)
print("\n--- Jogadores e novos salários atribuídos ---")
pd.options.display.float_format = '${:,.2f}'.format
display(df.loc[indices_nan, ['Name', 'Team', 'Position', 'Salary']])
```

```python
# ==============================================================================
# 39. Outlier Detection & Methodology Comparison (IQR Method)
# ==============================================================================

# 1. Obter valores de salário sem nulos
salaries = df['Salary'].dropna()

# 2. Computar Q1 (percentil 25), Q3 (percentil 75) e o IQR
Q1 = salaries.quantile(0.25)
Q3 = salaries.quantile(0.75)
IQR = Q3 - Q1

# 3. Definir os limites inferior e superior
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 4. Identificar e filtrar os outliers
outliers_iqr = df[(df['Salary'] < lower_bound) | (df['Salary'] > upper_bound)]

# 5. Apresentar os resultados calculados
print("--- Estatísticas do Método IQR ---")
print(f"Q1 (25th percentile): {Q1:,.2f}")
print(f"Q3 (75th percentile): {Q3:,.2f}")
print(f"IQR: {IQR:,.2f}")
print(f"Lower Bound: {lower_bound:,.2f}")
print(f"Upper Bound: {upper_bound:,.2f}")
print(f"\nTotal de outliers detetados: {len(outliers_iqr)}")

# 6. Mostrar os jogadores classificados como outliers (ordenados por salário)
cols = [col for col in ['Name', 'Team', 'Position', 'Salary'] if col in df.columns]
print("\n--- Jogadores com Salário Outlier ---")
display(outliers_iqr[cols].sort_values(by='Salary', ascending=False))
```

```python
# ==============================================================================
# 39. Outlier Detection & Methodology Comparison (b) Modified Z-Score Method)
# ==============================================================================

# 1. Obter valores de salário sem nulos
salaries = df['Salary'].dropna()

# 2. Calcular Mediana e o MAD (Median Absolute Deviation)
median_salary = salaries.median()
mad_salary = np.median(np.abs(salaries - median_salary))

# 3. Calcular o Modified Z-Score: 0.6745 * (xi - Median) / MAD
# Nota: se houver valores nulos em df['Salary'], mantemos o alinhamento de índices
df['Modified_ZScore'] = 0.6745 * (df['Salary'] - median_salary) / mad_salary

# 4. Definir o limiar (Threshold > 3.5) e filtrar os outliers
threshold = 3.5
outliers_mod_z = df[df['Modified_ZScore'].abs() > threshold]

# 5. Apresentar os resultados do método
print("--- Estatísticas do Método Modified Z-Score ---")
print(f"Mediana: {median_salary:,.2f}")
print(f"MAD (Median Absolute Deviation): {mad_salary:,.2f}")
print(f"Limiar (|Modified Z-Score| > {threshold})")
print(f"\nTotal de outliers detetados: {len(outliers_mod_z)}")

# 6. Mostrar os jogadores classificados como outliers (ordenados por salário)
cols = [col for col in ['Name', 'Team', 'Position', 'Salary', 'Modified_ZScore'] if col in df.columns]
print("\n--- Jogadores com Salário Outlier (Modified Z-Score) ---")
display(outliers_mod_z[cols].sort_values(by='Salary', ascending=False))
```

```python
# ==============================================================================
# 39. Outlier Detection & Methodology Comparison (c) Comparison & Visualization)
# ==============================================================================

# 1. Recalcular conjuntos de outliers para comparação direta
# IQR
salaries = df['Salary'].dropna()
Q1 = salaries.quantile(0.25)
Q3 = salaries.quantile(0.75)
IQR = Q3 - Q1
lower_iqr = Q1 - 1.5 * IQR
upper_iqr = Q3 + 1.5 * IQR
outliers_iqr_idx = set(df[(df['Salary'] < lower_iqr) | (df['Salary'] > upper_iqr)].index)

# Modified Z-Score
median_val = salaries.median()
mad_val = np.median(np.abs(salaries - median_val))
mod_z_scores = 0.6745 * (df['Salary'] - median_val) / mad_val
outliers_mod_z_idx = set(df[mod_z_scores.abs() > 3.5].index)

# 2. Operações de conjuntos (Comuns vs Exclusivos)
common_outliers = outliers_iqr_idx.intersection(outliers_mod_z_idx)
only_iqr = outliers_iqr_idx - outliers_mod_z_idx
only_mod_z = outliers_mod_z_idx - outliers_iqr_idx

print("--- Comparação entre Métodos de Outliers ---")
print(f"Total detetado por IQR: {len(outliers_iqr_idx)}")
print(f"Total detetado por Modified Z-Score: {len(outliers_mod_z_idx)}")
print(f"Outliers comuns a ambos os métodos: {len(common_outliers)}")
print(f"Outliers identificados APENAS pelo IQR: {len(only_iqr)}")
print(f"Outliers identificados APENAS pelo Modified Z-Score: {len(only_mod_z)}")

# Colunas para visualização
cols_show = [col for col in ['Name', 'Team', 'Position', 'Salary'] if col in df.columns]

if len(only_iqr) > 0:
    print("\n--- Jogadores identificados APENAS pelo IQR ---")
    display(df.loc[list(only_iqr), cols_show].sort_values(by='Salary', ascending=False))

if len(only_mod_z) > 0:
    print("\n--- Jogadores identificados APENAS pelo Modified Z-Score ---")
    display(df.loc[list(only_mod_z), cols_show].sort_values(by='Salary', ascending=False))

# 3. Criar uma coluna categórica de status para a visualização
df['Outlier_Class'] = 'Regular'
df.loc[list(common_outliers), 'Outlier_Class'] = 'Ambos'
df.loc[list(only_iqr), 'Outlier_Class'] = 'Apenas IQR'
df.loc[list(only_mod_z), 'Outlier_Class'] = 'Apenas Mod Z-Score'

# 4. Visualização Gráfica (Strip/Scatter Plot + Linhas de Corte)
plt.figure(figsize=(11, 6))

palette = {
    'Regular': 'gray',
    'Ambos': 'red',
    'Apenas IQR': 'dodgerblue',
    'Apenas Mod Z-Score': 'orange'
}

# Gráfico de dispersão com jitter horizontal para melhor legibilidade
sns.stripplot(
    data=df,
    x='Salary',
    hue='Outlier_Class',
    palette=palette,
    jitter=0.25,
    size=7,
    alpha=0.85
)

# Adicionar linhas de corte do IQR
plt.axvline(upper_iqr, color='dodgerblue', linestyle='--', linewidth=1.5, label=f'Limite Superior IQR ({upper_iqr:,.0f})')

# Adicionar linha de corte do Modified Z-Score (despejando xi para Mod_Z = 3.5)
cutoff_mod_z = median_val + (3.5 * mad_val / 0.6745)
plt.axvline(cutoff_mod_z, color='orange', linestyle=':', linewidth=1.5, label=f'Limite Superior Mod Z ({cutoff_mod_z:,.0f})')

plt.title('Comparação de Outliers Salariais: IQR vs Modified Z-Score')
plt.xlabel('Salário ($)')
plt.ylabel('Densidade / Dispersão')
plt.legend(title='Classificação', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
